```python
import pandas as pd

contracts = pd.read_csv("Supplier_Contracts_20260812.csv")

payments = pd.read_csv(
    "Vendor_Payments_(Purchase_Order_Summary)_20260812.csv",
    low_memory=False
)

print("Contracts:", contracts.shape)
print("Payments:", payments.shape)
```

```python
print("CONTRACTS SAMPLE")
display(contracts.head(3))

print("\nPAYMENTS SAMPLE")
display(payments.head(3))

print("\nCONTRACTS DATA TYPES")
print(contracts.dtypes)

print("\nPAYMENTS DATA TYPES")
print(payments.dtypes)
```

```python
print("MISSING VALUES — CONTRACTS")
print(contracts.isna().sum().sort_values(ascending=False).head(10))

print("\nMISSING VALUES — PAYMENTS")
print(payments.isna().sum().sort_values(ascending=False).head(10))

print("\nCONTRACT NUMBER COVERAGE")
print("Contracts with contract number:", contracts["Contract Number"].notna().sum())
print("Payments with contract number:", payments["Contract Number"].notna().sum())

print("\nUNIQUE CONTRACT NUMBERS")
print("Contracts:", contracts["Contract Number"].nunique())
print("Payments:", payments["Contract Number"].nunique())
```

```python
contract_amounts = ["Contract Awarded Amount", "Purchase Orders Outstanding", "Payments Made", "Remaining Contract Award Amount"]
payment_amounts = ["Vouchers Paid", "Vouchers Pending", "Encumbrance Balance", "Vouchers Pending Retainage"]

for col in contract_amounts:
    contracts[col] = contracts[col].str.replace("$", "", regex=False).str.replace(",", "", regex=False).str.strip().replace("", pd.NA).astype(float)

for col in payment_amounts:
    payments[col] = payments[col].str.replace("$", "", regex=False).str.replace(",", "", regex=False).str.strip().replace("", pd.NA).astype(float)

print("Contract amount columns:")
print(contracts[contract_amounts].dtypes)
print("\nPayment amount columns:")
print(payments[payment_amounts].dtypes)
```

```python
contracts["Contract Number"] = contracts["Contract Number"].apply(lambda x: str(int(x)) if pd.notna(x) else pd.NA)
payments["Contract Number"] = payments["Contract Number"].apply(lambda x: str(int(x)) if pd.notna(x) else pd.NA)

print("Contract ID type:", contracts["Contract Number"].dtype)
print("Payment ID type:", payments["Contract Number"].dtype)
print("Sample contract IDs:", contracts["Contract Number"].dropna().head().tolist())
```

```python
contract_counts = contracts["Contract Number"].value_counts()

print("Unique contract numbers:", contract_counts.size)
print("Contract numbers appearing more than once:", (contract_counts > 1).sum())
print("Maximum rows for one contract:", contract_counts.max())

print("\nMost repeated contract numbers:")
print(contract_counts.head(10))
```

```python
sample_contract = contracts[contracts["Contract Number"] == "1000003991"]

print("Rows for contract 1000003991:", len(sample_contract))
display(sample_contract[[
    "Contract Number",
    "Supplier Name (Prime Contractor)",
    "Supplier Name",
    "Department",
    "Contract Awarded Amount",
    "Payments Made",
    "Remaining Contract Award Amount"
]].head(20))
```

```python
contract_view = sample_contract[[
    "Contract Awarded Amount",
    "Payments Made",
    "Remaining Contract Award Amount"
]].sum()

print("Total awarded amount:", contract_view["Contract Awarded Amount"])
print("Total payments made:", contract_view["Payments Made"])
print("Total remaining amount:", contract_view["Remaining Contract Award Amount"])
```

```python
contract_payments = payments[payments["Contract Number"].notna()].copy()

contract_spend = contract_payments.groupby("Contract Number").agg(
    FY2024_Spend=("Vouchers Paid", lambda x: x[contract_payments.loc[x.index, "Fiscal Year"].eq(2024)].sum()),
    FY2025_Spend=("Vouchers Paid", lambda x: x[contract_payments.loc[x.index, "Fiscal Year"].eq(2025)].sum()),
    Total_Spend=("Vouchers Paid", "sum"),
    Purchase_Orders=("Purchase Order", "nunique")
).reset_index()

print("Contract-level payment summary:", contract_spend.shape)
display(contract_spend.head())
```

```python
print("Payment rows with contract numbers:", len(contract_payments))
print("Unique contracts:", contract_payments["Contract Number"].nunique())

print("\nTotal vouchers paid from payment records:",
      contract_payments["Vouchers Paid"].sum())

print("Total spend in contract summary:",
      contract_spend["Total_Spend"].sum())

print("\nDifference:",
      contract_payments["Vouchers Paid"].sum() - contract_spend["Total_Spend"].sum())
```

```python
payment_contract_ids = set(contract_spend["Contract Number"].dropna())
contract_ids = set(contracts["Contract Number"].dropna())

matched_ids = payment_contract_ids.intersection(contract_ids)

print("Contracts with payment activity:", len(payment_contract_ids))
print("Matching contract records:", len(matched_ids))
print("Unmatched contract IDs:", len(payment_contract_ids - contract_ids))
print("Match rate:", round(len(matched_ids) / len(payment_contract_ids) * 100, 2), "%")
```

```python
contract_analysis = contracts.merge(contract_spend, on="Contract Number", how="left")
contract_analysis["Total_Spend"] = contract_analysis["Total_Spend"].fillna(0)
contract_analysis["FY2024_Spend"] = contract_analysis["FY2024_Spend"].fillna(0)
contract_analysis["FY2025_Spend"] = contract_analysis["FY2025_Spend"].fillna(0)
contract_analysis["Purchase_Orders"] = contract_analysis["Purchase_Orders"].fillna(0)

print("Rows in analysis table:", len(contract_analysis))
print("Columns:", len(contract_analysis.columns))
display(contract_analysis.head())
```

```python
contract_level = contracts.groupby("Contract Number", dropna=False).agg(
    Contract_Title=("Contract Title", "first"),
    Department=("Department", "first"),
    Contract_Type=("Contract Type", "first"),
    Prime_Contractor=("Supplier Name (Prime Contractor)", "first"),
    Awarded_Amount=("Contract Awarded Amount", "sum"),
    Payments_Made=("Payments Made", "sum"),
    Remaining_Amount=("Remaining Contract Award Amount", "sum")
).reset_index()

contract_level = contract_level.merge(
    contract_spend,
    on="Contract Number",
    how="left"
)

for col in ["FY2024_Spend", "FY2025_Spend", "Total_Spend", "Purchase_Orders"]:
    contract_level[col] = contract_level[col].fillna(0)

print("Contract-level table:", contract_level.shape)
display(contract_level.head())
```

```python
contract_level["Utilization_Rate"] = (contract_level["Payments_Made"] / contract_level["Awarded_Amount"] * 100).fillna(0).clip(upper=100)

print("Average utilization:", round(contract_level["Utilization_Rate"].mean(), 2), "%")
print("Median utilization:", round(contract_level["Utilization_Rate"].median(), 2), "%")

contract_level[["Contract Number", "Awarded_Amount", "Payments_Made", "Remaining_Amount", "Utilization_Rate"]].head(10)
```

```python
contract_level["Utilization_Band"] = pd.cut(contract_level["Utilization_Rate"], bins=[-1, 25, 50, 75, 100], labels=["0-25%", "25-50%", "50-75%", "75-100%"])
print(contract_level["Utilization_Band"].value_counts().sort_index())
```

```python
underused = contract_level[(contract_level["Utilization_Rate"] < 25) & (contract_level["Awarded_Amount"] > 100000)].copy()
underused = underused.sort_values("Awarded_Amount", ascending=False)

print("High-value contracts with utilization below 25%:", len(underused))
underused[["Contract Number", "Contract_Title", "Department", "Prime_Contractor", "Awarded_Amount", "Payments_Made", "Remaining_Amount", "Utilization_Rate"]].head(15)
```

```python
print("Total awarded value:", round(underused["Awarded_Amount"].sum(), 2))
print("Total payments made:", round(underused["Payments_Made"].sum(), 2))
print("Total remaining value:", round(underused["Remaining_Amount"].sum(), 2))
print("Share of all awarded value:", round(underused["Awarded_Amount"].sum() / contract_level["Awarded_Amount"].sum() * 100, 2), "%")
```

```python
department_exposure = underused.groupby("Department").agg(Contracts=("Contract Number", "nunique"),Awarded_Value=("Awarded_Amount", "sum"),Remaining_Value=("Remaining_Amount", "sum")
).sort_values("Remaining_Value", ascending=False).reset_index()

department_exposure["Remaining_Share"] = department_exposure["Remaining_Value"] / underused["Remaining_Amount"].sum() * 100

department_exposure.head(10)
```

```python
supplier_summary = contract_level.groupby("Prime_Contractor").agg(Contracts=("Contract Number", "nunique"), Awarded_Value=("Awarded_Amount", "sum"), Payments_Made=("Payments_Made", "sum")).sort_values("Awarded_Value", ascending=False).reset_index()
supplier_summary.head(15)
```

```python
supplier_summary["Over_Award"] = supplier_summary["Payments_Made"] > supplier_summary["Awarded_Value"]
print("Suppliers with payments above awarded value:", supplier_summary["Over_Award"].sum())
print("Total suppliers:", len(supplier_summary))
```

```python
supplier_spend = payments[payments["Supplier & Other Non-Supplier Payees"].notna()].groupby("Supplier & Other Non-Supplier Payees").agg(Total_Spend=("Vouchers Paid", "sum"), Purchase_Orders=("Purchase Order", "nunique")).sort_values("Total_Spend", ascending=False).reset_index()

print("Suppliers with payment activity:", len(supplier_spend))
supplier_spend.head(15)
```

```python
object_spend = payments.groupby("Object").agg(Total_Spend=("Vouchers Paid", "sum"), Purchase_Orders=("Purchase Order", "nunique")).sort_values("Total_Spend", ascending=False).reset_index()
object_spend.head(20)
```

```python
procurement_objects = ["Professional/Specialized Svcs", "Rent/Lease-Building/Structure", "Community Based Org Srvcs", "Bldg: Structures/Improvements", "Membership Fees", "Insurance", "Other Current Expenses"]
procurement_payments = payments[payments["Object"].isin(procurement_objects)].copy()
print("Procurement-focused payment rows:", len(procurement_payments))
print("Procurement-focused spend:", round(procurement_payments["Vouchers Paid"].sum(), 2))
```

```python
procurement_department = procurement_payments.groupby("Department").agg(Total_Spend=("Vouchers Paid", "sum"), Purchase_Orders=("Purchase Order", "nunique")).sort_values("Total_Spend", ascending=False).reset_index()
procurement_department.head(15)
```

```python
object_summary = payments.groupby("Object").agg(Total_Spend=("Vouchers Paid", "sum"), Transactions=("Vouchers Paid", "size")).sort_values("Total_Spend", ascending=False).reset_index()
object_summary.head(40)
```

```python
procurement_objects = ["Professional/Specialized Svcs", "Rent/Lease-Building/Structure", "Community Based Org Srvcs", "Bldg: Structures/Improvements", "Membership Fees", "Insurance", "Equipment Purchase", "Training", "Maint Svcs-Building & Structur", "Maint Svcs-Equipment", "Building/Construction Supplies", "Other Materials/Supplies", "Equipment Mntc Supplies", "Rents/Leases-Equipment", "Fuels/Lubricants", "Utilities", "Travel", "Entertainment and Promotion"]

procurement_payments = payments[payments["Object"].isin(procurement_objects)].copy()

print("Procurement transactions:", len(procurement_payments))
print("Procurement spend:", round(procurement_payments["Vouchers Paid"].sum(), 2))
print("Share of all payment spend:", round(procurement_payments["Vouchers Paid"].sum() / payments["Vouchers Paid"].sum() * 100, 2), "%")
```

```python
supplier_spend = procurement_payments.groupby("Supplier & Other Non-Supplier Payees").agg(Total_Spend=("Vouchers Paid", "sum"), Purchase_Orders=("Purchase Order", "nunique")).sort_values("Total_Spend", ascending=False).reset_index()

print("Suppliers:", len(supplier_spend))
supplier_spend.head(15)
```

```python
supplier_spend["Spend_Share"] = supplier_spend["Total_Spend"] / supplier_spend["Total_Spend"].sum() * 100
supplier_spend["Cumulative_Share"] = supplier_spend["Spend_Share"].cumsum()

top_10_share = supplier_spend.head(10)["Total_Spend"].sum() / supplier_spend["Total_Spend"].sum() * 100
top_20_share = supplier_spend.head(20)["Total_Spend"].sum() / supplier_spend["Total_Spend"].sum() * 100

print("Top 10 supplier spend share:", round(top_10_share, 2), "%")
print("Top 20 supplier spend share:", round(top_20_share, 2), "%")
```
